# 2. Data Clear

## Clear missing values from customerId and Description

```python
df = df.dropna(subset=["CustomerID", "Description"])
df.info()
```

## Find the duplicates

```python
df.drop_duplicates()
```

## Find Canceled orders starts with 'C'

```python
df = df[~df["InvoiceNo"].astype(str).str.startswith("C", na=False)]
df
```

## Prices and Quantity with 0 and Negative values

```python
mask = (df["Quantity"] != 0) | (df["UnitPrice"] < 0)
df = df[mask] 
df.info()
```

## Date to datetime type


```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df.info()
```

## New column - *Revenue* (Quantity + UnitPrice)

```python
df["Revenue"] = round(df["Quantity"] * df["UnitPrice"], 2)
df
```

## Output rows with Revenue > 10

```python
df.loc[df["Revenue"] > 10] 
```

## Output Id, Description where Revenue > 10

```python
df.loc[df["Revenue"] > 10, ["InvoiceNo", "Description"]] 
```

## Add Year, Month and Hour columns

```python
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Hour"] = df["InvoiceDate"].dt.hour
df
```

## Save table to CSV

```python
df.to_csv("../data/processed/OnleineRetail_clear1.csv")
```
